```python
df_clean = df.copy()
```

```python
missing = df_clean.isnull().sum()

missing = missing[missing > 0]

missing_percentage = (
    missing / len(df_clean) * 100
).round(2)

missing_df = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_percentage
})

missing_df.sort_values(
    by="Missing_Percentage",
    ascending=False
)
```

```python
df_clean["Customer_Age"] = (
    df_clean["Customer_Age"]
    .fillna(df_clean["Customer_Age"].median())
)
```

```python
df_clean["Customer_Gender"] = (
    df_clean["Customer_Gender"]
    .fillna("Unknown")
)
```

```python
df_clean["City"] = df_clean["City"].fillna("Unknown")

df_clean["Area"] = df_clean["Area"].fillna("Unknown")

df_clean["Cuisine_Type"] = (
    df_clean["Cuisine_Type"].fillna("Unknown")
)
```

```python
df_clean["Payment_Mode"] = (
    df_clean["Payment_Mode"]
    .fillna("Unknown")
)
```

```python
df_clean["Order_Value"] = (
    df_clean["Order_Value"]
    .fillna(df_clean["Order_Value"].median())
)
```

```python
df_clean["Discount_Applied"] = (
    df_clean["Discount_Applied"]
    .fillna(df_clean["Discount_Applied"].median())
)
```

```python
df_clean["Final_Amount"] = (
    df_clean["Order_Value"]
    - df_clean["Discount_Applied"]
)
```

```python
df_clean["Distance_km"] = (
    df_clean["Distance_km"]
    .fillna(df_clean["Distance_km"].median())
)
```

```python
df_clean["Delivery_Time_Min"] = (
    df_clean["Delivery_Time_Min"]
    .fillna(df_clean["Delivery_Time_Min"].median())
)
```

```python
delivered_mask = df_clean["Order_Status"] == "Delivered"

median_rating = df_clean.loc[
    delivered_mask,
    "Delivery_Rating"
].median()

df_clean.loc[
    delivered_mask & df_clean["Delivery_Rating"].isna(),
    "Delivery_Rating"
] = median_rating
```

```python
df_clean.loc[
    df_clean["Order_Status"] == "Delivered",
    "Cancellation_Reason"
] = "Not Applicable"
```

```python
df_clean.loc[
    (df_clean["Order_Status"] == "Cancelled") &
    (df_clean["Cancellation_Reason"].isna()),
    "Cancellation_Reason"
] = "Unknown"
```

```python
df_clean["Peak_Hour"] = (
    df_clean["Peak_Hour"]
    .fillna("Unknown")
)
```

```python
df_clean["Order_Time"] = (
    df_clean["Order_Time"].fillna("0:00")
)
```

```python
df_clean["Order_Date"] = pd.to_datetime(
    df_clean["Order_Date"],
    errors="coerce"
)
```

```python
monthly_df = df_clean.dropna(
    subset=["Order_Date"]
)
```
