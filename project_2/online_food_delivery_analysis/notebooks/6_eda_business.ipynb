```python
monthly_revenue = (
    df
    .dropna(subset=["Order_Date"])
    .groupby(
        ["Order_Month_Number", "Order_Month"]
    )
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Orders=("Order_ID", "count")
    )
    .reset_index()
    .sort_values("Order_Month_Number")
)

monthly_revenue
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_revenue["Order_Month"],
    monthly_revenue["Total_Revenue"],
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Revenue")
plt.title("Monthly Revenue Trend")
plt.xticks(rotation=45)

plt.show()
```

```python
city_delivery = (
    df[
        df["City"] != "Unknown"
    ]
    .groupby("City")
    .agg(
        Average_Delivery_Time=(
            "Delivery_Time_Min",
            "mean"
        ),
        Median_Delivery_Time=(
            "Delivery_Time_Min",
            "median"
        ),
        Total_Orders=(
            "Order_ID",
            "count"
        )
    )
    .reset_index()
    .sort_values(
        "Average_Delivery_Time",
        ascending=False
    )
)

city_delivery
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    city_delivery["City"],
    city_delivery["Average_Delivery_Time"]
)

plt.xlabel("City")
plt.ylabel("Average Delivery Time (Minutes)")
plt.title("Average Delivery Time by City")

plt.xticks(rotation=45)

plt.show()
```

```python
restaurant_rating = (
    df
    .groupby("Restaurant_Name")
    .agg(
        Average_Rating=(
            "Restaurant_Rating",
            "mean"
        ),
        Total_Orders=(
            "Order_ID",
            "count"
        )
    )
    .reset_index()
)
```

```python
restaurant_rating = restaurant_rating[
    restaurant_rating["Total_Orders"] >= 50
]
```

```python
top_restaurants = (
    restaurant_rating
    .sort_values(
        ["Average_Rating", "Total_Orders"],
        ascending=[False, False]
    )
    .head(10)
)

top_restaurants
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_restaurants["Restaurant_Name"],
    top_restaurants["Average_Rating"]
)

plt.xlabel("Average Restaurant Rating")
plt.ylabel("Restaurant")
plt.title("Top 10 Rated Restaurants")

plt.show()
```

```python
restaurant_cancel = (
    df
    .groupby("Restaurant_Name")
    .agg(
        Total_Orders=("Order_ID", "count"),
        Cancelled_Orders=("Is_Cancelled", "sum")
    )
    .reset_index()
)
```

```python
restaurant_cancel["Cancellation_Rate"] = (
    restaurant_cancel["Cancelled_Orders"]
    / restaurant_cancel["Total_Orders"]
    * 100
)
```

```python
restaurant_cancel = restaurant_cancel[
    restaurant_cancel["Total_Orders"] >= 50
]
```

```python
highest_cancellation = (
    restaurant_cancel
    .sort_values(
        "Cancellation_Rate",
        ascending=False
    )
    .head(10)
)

highest_cancellation
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    highest_cancellation["Restaurant_Name"],
    highest_cancellation["Cancellation_Rate"]
)

plt.xlabel("Cancellation Rate (%)")
plt.ylabel("Restaurant")
plt.title("Restaurants with Highest Cancellation Rate")

plt.show()
```

```python
cuisine_performance = (
    df[
        df["Cuisine_Type"] != "Unknown"
    ]
    .groupby("Cuisine_Type")
    .agg(
        Total_Orders=(
            "Order_ID",
            "count"
        ),
        Total_Revenue=(
            "Revenue",
            "sum"
        ),
        Average_Order_Value=(
            "Order_Value",
            "mean"
        ),
        Average_Delivery_Time=(
            "Delivery_Time_Min",
            "mean"
        ),
        Average_Restaurant_Rating=(
            "Restaurant_Rating",
            "mean"
        ),
        Cancellation_Rate=(
            "Is_Cancelled",
            "mean"
        )
    )
    .reset_index()
)
```

```python
cuisine_performance["Cancellation_Rate"] *= 100
```

```python
peak_analysis = (
    df[
        df["Peak_Hour"] != "Unknown"
    ]
    .groupby("Peak_Hour")
    .agg(
        Total_Orders=(
            "Order_ID",
            "count"
        ),
        Average_Order_Value=(
            "Order_Value",
            "mean"
        ),
        Average_Delivery_Time=(
            "Delivery_Time_Min",
            "mean"
        ),
        Cancellation_Rate=(
            "Is_Cancelled",
            "mean"
        ),
        Revenue=(
            "Revenue",
            "sum"
        )
    )
    .reset_index()
)
```

```python
peak_analysis["Cancellation_Rate"] *= 100
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    peak_analysis["Peak_Hour"],
    peak_analysis["Total_Orders"]
)

plt.xlabel("Period")
plt.ylabel("Number of Orders")
plt.title("Peak vs Non-Peak Demand")

plt.show()
```

```python
payment_analysis = (
    df[
        df["Payment_Mode"] != "Unknown"
    ]
    .groupby("Payment_Mode")
    .agg(
        Total_Orders=("Order_ID", "count"),
        Total_Revenue=("Revenue", "sum"),
        Average_Order_Value=("Order_Value", "mean")
    )
    .reset_index()
)
```

```python
payment_analysis["Order_Percentage"] = (
    payment_analysis["Total_Orders"]
    / payment_analysis["Total_Orders"].sum()
    * 100
)
```

```python
payment_analysis.sort_values(
    "Total_Orders",
    ascending=False
)
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    payment_analysis["Payment_Mode"],
    payment_analysis["Total_Orders"]
)

plt.xlabel("Payment Mode")
plt.ylabel("Number of Orders")
plt.title("Payment Mode Preference")

plt.show()
```

```python
cancelled_df = df[
    df["Order_Status"] == "Cancelled"
]
```

```python
cancellation_reasons = (
    cancelled_df[
        "Cancellation_Reason"
    ]
    .value_counts()
    .reset_index()
)

cancellation_reasons.columns = [
    "Cancellation_Reason",
    "Count"
]

cancellation_reasons
```

```python
cancellation_reasons["Percentage"] = (
    cancellation_reasons["Count"]
    / cancellation_reasons["Count"].sum()
    * 100
)

cancellation_reasons
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    cancellation_reasons["Cancellation_Reason"],
    cancellation_reasons["Count"]
)

plt.xlabel("Cancellation Reason")
plt.ylabel("Number of Orders")
plt.title("Cancellation Reasons")

plt.xticks(rotation=30)

plt.show()
```
