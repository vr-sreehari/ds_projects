```python
def categorical_summary(df, column):
    count = df[column].value_counts(dropna=False)

    percentage = (
        df[column]
        .value_counts(
            normalize=True,
            dropna=False
        )
        * 100
    ).round(2)

    result = pd.DataFrame({
        "Count": count,
        "Percentage": percentage
    })

    return result
```

```python
categorical_summary(
    df_clean,
    "City"
)
```

```python
categorical_summary(
    df_clean,
    "Payment_Mode"
)
```

```python
categorical_summary(
    df_clean,
    "Order_Status"
)
```
